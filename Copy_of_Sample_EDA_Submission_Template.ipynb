Handle missng values

````python

# Since many event-related columns contain missing values,
# replace NaN values with "Not Available"

event_columns = [
    'home_goal', 'away_goal', 'home_goal_long', 'away_goal_long',
    'home_own_goal', 'away_own_goal',
    'home_penalty_goal', 'away_penalty_goal',
    'home_penalty_miss_long', 'away_penalty_miss_long',
    'home_penalty_shootout_goal_long', 'away_penalty_shootout_goal_long',
    'home_penalty_shootout_miss_long', 'away_penalty_shootout_miss_long',
    'home_red_card', 'away_red_card',
    'home_yellow_red_card', 'away_yellow_red_card',
    'home_yellow_card_long', 'away_yellow_card_long',
    'home_substitute_in_long', 'away_substitute_in_long',
    'Notes'
]

for column in event_columns:
    if column in df.columns:
        df[column].fillna("Not Available", inplace=True)

print("Missing values handled successfully.")
````

#### Chart - 8

````python

# Chart 8 : Top 10 Teams with the Most FIFA World Cup Wins

# Identify winners of each match
home_wins = df[df["home_score"] > df["away_score"]]["home_team"]
away_wins = df[df["away_score"] > df["home_score"]]["away_team"]

# Combine all winning teams
all_wins = pd.concat([home_wins, away_wins])

# Count wins for each team
top_wins = all_wins.value_counts().head(10)

# Create figure
plt.figure(figsize=(14,7))

# Create colorful horizontal bar chart
ax = sns.barplot(
    x=top_wins.values,
    y=top_wins.index,
    palette="turbo",
    edgecolor="black"
)

# Add value labels
for container in ax.containers:
    ax.bar_label(container, fontsize=11, fontweight="bold")

# Title and labels
plt.title(
    "Top 10 Teams with the Most FIFA World Cup Match Wins",
    fontsize=18,
    fontweight="bold",
    pad=15
)

plt.xlabel("Number of Wins", fontsize=13)
plt.ylabel("Team", fontsize=13)

# Grid
plt.grid(axis='x', linestyle='--', alpha=0.4)

plt.tight_layout()
plt.show()
````

````python
# ==========================================================
# Chart 9 : Top 10 Teams with the Most Goals Scored
# ==========================================================

# Goals scored as Home Team
home_goals = df.groupby("home_team")["home_score"].sum()

# Goals scored as Away Team
away_goals = df.groupby("away_team")["away_score"].sum()

# Total goals scored by each team
total_goals = (home_goals + away_goals).sort_values(ascending=False).head(10)

# Sort for better horizontal visualization
total_goals = total_goals.sort_values()

# Create figure
plt.figure(figsize=(14,7))

# Create colorful horizontal bar chart
ax = sns.barplot(
    x=total_goals.values,
    y=total_goals.index,
    palette="flare",
    edgecolor="black",
    linewidth=1.2
)

# Add value labels
for container in ax.containers:
    ax.bar_label(
        container,
        fontsize=11,
        fontweight="bold",
        padding=5
    )

# Title and Labels
plt.title(
    "Top 10 Teams with the Most Goals Scored in FIFA World Cup History",
    fontsize=18,
    fontweight="bold",
    pad=15
)

plt.xlabel("Total Goals Scored", fontsize=13)
plt.ylabel("National Team", fontsize=13)

# Grid
plt.grid(axis='x', linestyle='--', alpha=0.4)

plt.tight_layout()
plt.show()
````

#### Chart - 10

````python

# Top 10 Highest Scoring FIFA World Cup Matches


# Create a new column for total goals scored in a match
df["Total Goals"] = df["home_score"] + df["away_score"]

# Select top 10 highest scoring matches
top_matches = (
    df.nlargest(10, "Total Goals")
    .copy()
)

# Create a readable match label
top_matches["Match"] = (
    top_matches["home_team"] + " vs " + top_matches["away_team"]
)

# Sort for better horizontal visualization
top_matches = top_matches.sort_values("Total Goals")

# Create figure
plt.figure(figsize=(15,8))

# Plot
ax = sns.barplot(
    data=top_matches,
    x="Total Goals",
    y="Match",
    palette="coolwarm",
    edgecolor="black",
    linewidth=1.2
)

# Add labels
for container in ax.containers:
    ax.bar_label(
        container,
        fontsize=11,
        fontweight="bold",
        padding=5
    )

# Title
plt.title(
    "Top 10 Highest Scoring FIFA World Cup Matches",
    fontsize=18,
    fontweight="bold",
    pad=15
)

plt.xlabel("Total Goals Scored", fontsize=13)
plt.ylabel("Match", fontsize=13)

plt.grid(axis="x", linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
````
